# 01 - Phase 1 Data Exploration Notebook
This notebook explores the clean, integrated time-series datasets created in Phase 1 of the **AI-Enabled Predictive Price Intelligence System**.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Load clean datasets
rice_prod = pd.read_parquet('../data/processed/rice_production.parquet')
rice_mkt = pd.read_parquet('../data/processed/rice_market.parquet')
paddy_mkt = pd.read_parquet('../data/processed/paddy_market.parquet')
weather = pd.read_parquet('../data/processed/weather.parquet')
stock = pd.read_parquet('../data/processed/stock.parquet')
fe_dataset = pd.read_parquet('../data/processed/feature_engineered_modelling_dataset.parquet')

print('Rice Production Shape:', rice_prod.shape)
print('Rice Market Shape:', rice_mkt.shape)
print('Paddy Market Shape:', paddy_mkt.shape)
print('Feature Engineered Dataset Shape:', fe_dataset.shape)
```

## 1. Rice Production by State and Financial Year

```python
rice_prod.groupby(['financial_year'])['production_mt'].sum().plot(kind='bar', color='skyblue')
plt.title('All India Total Rice Production (MT) across Financial Years')
plt.ylabel('Production (MT)')
plt.show()
```

## 2. Market Modal Price Trends (Rice vs Paddy)

```python
print('Rice Market Price Stats (₹/Qtl):')
print(rice_mkt['price_rs_per_qtl'].describe())

print('\nPaddy Market Price Stats (₹/Qtl):')
print(paddy_mkt['price_rs_per_qtl'].describe())
```

## 3. Buffer Stock vs Buffer Norms

```python
stock.plot(x='date', y=['rice_stock_mt', 'buffer_norm_mt'], figsize=(10, 4))
plt.title('FCI Central Pool Rice Stock vs Quarterly Buffer Norms')
plt.ylabel('Metric Tonnes (MT)')
plt.show()
```
